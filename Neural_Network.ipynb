# Building Neural Network From Scratch Part 1: Coding Neurons And Layers

### Coding Our First Neuron: 3 INPUTS

```python
inputs = [1, 2, 3]
weights = [0.2, 0.8, -0.5]
bias = 2

outputs = (inputs[0]*weights[0]+inputs[1]*weights[1]+inputs[2]*weights[2] + bias)

print(outputs)
```

### Coding Our second NEURON: 4 INPUTS

```python
inputs = [1.0, 2.0, 3.0, 2.5]
weights = [0.2, 0.8, -0.5, 1.0]
bias = 2.0
outputs = (inputs[0]*weights[0]+
           inputs[0]*weights[0]+
           inputs[0]*weights[0]+
           inputs[0]*weights[0] + bias)
print(outputs)
```

### Layer of Neurons

```python
inputs = [1, 2, 3, 2.5]

weights = [[0.2, 0.8, -0.5, 1],
          [0.5, -0.91, 0.26, -0.5],
          [-0.26,-0.27, 0.17, 0.87]]

weights1 = weights[0]  # List of weights associated with the first neuron : w11, w12, w13, w14
weights2 = weights[1]  # List of weights associated with the second neuron : w21, w22, w23, w24
weights3 = weights[2]  # List of weights associated with the third neuron : w31, w32, w33, w34


biases = [2, 3, 0.5]

bias1 = 2
bias2 = 3
bias3 = 0.5

outputs = [
    # Neurons 1:
    inputs[0]*weights1[0]+
    inputs[1]*weights1[1]+
    inputs[2]*weights1[2]+
    inputs[3]*weights1[3]+ bias1,

    # Neurons 2:
    inputs[0]*weights2[0]+
    inputs[1]*weights2[1]+
    inputs[2]*weights2[2]+
    inputs[3]*weights2[3]+ bias2,

    # Neuron 3:
    
    inputs[0]*weights3[0]+
    inputs[1]*weights3[1]+
    inputs[2]*weights3[2]+
    inputs[3]*weights3[3]+ bias3
]


print(outputs)
```

### Using Loops for better and easier coding

```python
inputs = [1.1, 2.2, 3.3]

## List of weights
weights = [[2,0,0],
          [2,2,0],
          [2,2,2]]

# list of biases
biases = [0,0,1]

# Output of current layer

layer_outputs = []

# For each neuron
for neuron_weights, neuron_bias in zip(weights, biases):
    # Zeroed output of given neuron
   neuron_output = 0
   # For each input and weight to the neuron
   for n_input, weight in zip(inputs, neuron_weights):
    # Multiply this inputs by associated weight
    # And add to the neuron's output variable
       neuron_output += n_input*weight   # w11*x1 + w12*x2 + w13*x3 + w14*x4
    # Add bias 
   neuron_output += neuron_bias ### w11*x1 + w12*x2 + w13*x3 + w14*x4 B1
 # Put neuron's results to the layers output list
   layer_outputs.append(neuron_output)
print(layer_outputs)  
```

```python
inputs = [1, 2, 3, 2.5]

# List of weights
weights = [
    [0.2, 0.8, -0.5, 1],
    [0.5, -0.91, 0.26, -0.5],
    [-0.26, -0.27, 0.17, 0.87]
]

# List of biases
biases = [2, 3, 0.5]

# Output of current layer
layer_outputs = []

# For each neuron
for neuron_weights, neuron_bias in zip(weights, biases):
    # Zeroed output of given neuron
    neuron_output = 0

    # For each input and weight to the neuron
    for n_input, weight in zip(inputs, neuron_weights):
        # Multiply input by weight and add to output
        neuron_output += n_input * weight

    # Add bias
    neuron_output += neuron_bias

    # Append neuron's output
    layer_outputs.append(neuron_output)

print(layer_outputs)
```
